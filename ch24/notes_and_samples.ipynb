# Parser combinators

**Parser combinators** are higher-order functions that take parsers as input and return a new parser as output. In computer science, a **parser** is a function that accepts serialized input—such as a sequence of characters or bytes—and transforms it into a structured datatype, which often takes the form of a tree. Combinators allow developers to glue these individual parsers together in a modular fashion to handle complex rules and recursion.

### Core Concepts and Structure
In Haskell, parser combinators leverage standard typeclasses such as **Applicative**, **Monad**, and **Alternative** to treat parsers like values that can be easily combined.

*   **The Cursor Analogy:** The process of parsing is often compared to moving a cursor across a linear stream of text. As a parser succeeds, it "consumes" the input and moves the cursor forward.
*   **State-like Behavior:** Although implementations vary, most Haskell parsing libraries have an interface that behaves like the **State monad**. The "state" being managed is the current position of the cursor in the input stream. A simplified representation of a parser type is `type Parser a = String -> Maybe (a, String)`, where it takes a string, produces a result (or fails with `Nothing`), and returns the remaining unconsumed string.

### Common Operations and Combinators
Combinators provide the building blocks for creating sophisticated parsers from simpler ones:
*   **Sequencing (`>>`):** Used to run parsers in a specific order, where the effect of moving the cursor is preserved even if the intermediate result is discarded.
*   **Choice (`<|>`):** Part of the `Alternative` typeclass, this allows for disjunction, where the program attempts one parser and, if it fails, moves to the next.
*   **Repetition (`some` and `many`):** These combinators allow for parsing "one or more" (`some`) or "zero or more" (`many`) instances of a pattern.
*   **Backtracking (`try`):** This is used when a parser might fail after consuming some input; `try` returns the cursor to its original starting position so another parser (often the second half of a `<|>` choice) can attempt to parse the same input.
*   **Tokenization:** Many libraries include tokenizers that group raw characters into "chunkier" structures like words or symbols, often automatically handling noise like whitespace and newlines.

### Haskell's Parsing Ecosystem
Haskell features several prominent libraries for working with parser combinators:
*   **Trifecta:** Known for providing highly readable and helpful error messages that pinpoint exactly where a parse failed.
*   **Parsec and Megaparsec:** Widely used and well-documented libraries that serve as a standard for many Haskell projects.
*   **Attoparsec:** Specialized for high-speed parsing, making it a common choice for performance-critical production environments.

By using these libraries, programmers can avoid the "mind-bending backflips" required to process unstructured text, instead converting it into **meaningful structures** that the rest of the program can use reliably.

3. Try writing a Parser that does what `string` does, but using `char`.

```haskell
string' :: String -> Parser String
string' "" = return ""
string' (c:cs) = do
  x <- char c
  xs <- string' cs
  return (x:xs)
```
